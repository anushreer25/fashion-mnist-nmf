## 3. NMF Parameter Selection

### 3.1. Elbow Plot
|
Construct a NMF elbow plot (see case study 1 in this assignment #5 for how we did this). This plot should plot the optimal objective function value for 45 NMF clusterings, where each clustering has k = 1, 2, ..., 45 clusters. Each clustering should be created with a random state of 437.

```python
from warnings import simplefilter
from sklearn.exceptions import ConvergenceWarning
simplefilter('ignore', category=ConvergenceWarning)

opt_values=[]

for k in range(1,46):
    nmf=NMF(n_components=k, random_state=437)
    nmf.fit(X)
    opt_values.append(nmf.reconstruction_err_)

plt.plot(range(1,46), opt_values)
plt.title('NMF elbow plot')
plt.xlabel('Cluster Number k')
plt.ylabel('NMF optimal objective function value')
plt.show()
```

## 4. NMF Clustering

Let's first use $k=10$ clusters because this corresponds to about how many clusters are suggested by the t-SNE plot (of the images) and we have 10 image classes.

### 4.1. NMF Clustering 

Cluster your dataset using NMF with k=12 clusters. Use a random state of 437. If your dataset is $X$ and you approximate $X\approx W\cdot H$ in your decomposition, display your W and H matrices in two dataframes.

```python
nmf=NMF(n_components=12, random_state=437)
W=nmf.fit_transform(X)
H=nmf.components_
```

```python
df_W=pd.DataFrame(W, columns=['image_cluster_'+str(k) for k in range(0,12)])
df_W.round(4)
```

```python
df_H=pd.DataFrame(H, columns=df_copy.columns)
df_H.round(4)
```

### 4.2. Display Results

Plot your t-SNE plot ten times: each time color coding the points by one of the columns in W.

```python
fig, ax = plt.subplots(2,5, figsize=(20,10))
plt.suptitle('NMF cclustering', fontsize=18)
for i in range(0,10):
    df_combo['Cluster %s Membership Score'%str(i)]=df_W.iloc[:,i]

    points=plt.scatter(df_combo['x_projected'], df_combo['y_projected'],
                c=df_combo['Cluster %s Membership Score'%str(i)],
                s=30,cmap='BuGn')
    ax[int(np.floor(i/5)), int(i%5)].scatter(df_combo['x_projected'], df_combo['y_projected'],
    c=df_combo['Cluster %s Membership Score'%str(i)],
    s=30,cmap='BuGn')

    fig.colorbar(points,ax=ax[int(np.floor(i/5)),int(i%5)])
    ax[int(np.floor(i/5)),int(i%5)].set_title('Cluster weights in column %s of W'%(str(i)))
plt.show()
```

### 4.3. Image Cluster Membership Matrix

Create a cluster membership matrix for for the images, using the W matrix.

```python
cm_scores_W=W/np.sum(W, axis=1, keepdims=True)
df_cm_scores_W = pd.DataFrame(cm_scores_W, columns=df_W.columns) 
```

```python
df_cm_scores_W
```

### 4.4. Interpretation

To what extent does the 28th image in the dataset belong to each of the 10 image clusters?

```python
df_cm_scores_W.loc[27]
```

### 4.5. Displaying the Cluster Membership Values

Plot your t-SNE plot 12 times: each time color coding the points by the ten cluster membership score values.

```python
fig, ax = plt.subplots(2,6, figsize=(20,10))
plt.suptitle('NMF cclustering', fontsize=18)
for i in range(0,12):
    df_combo['Cluster %s Membership Score'%str(i)]=df_cm_scores_W.iloc[:,i]

    points=plt.scatter(df_combo['x_projected'], df_combo['y_projected'],
                c=df_combo['Cluster %s Membership Score'%str(i)],
                s=30,cmap='BuGn')
    ax[int(np.floor(i/6)), int(i%6)].scatter(df_combo['x_projected'], df_combo['y_projected'],
    c=df_combo['Cluster %s Membership Score'%str(i)],
    s=30,cmap='BuGn')

    fig.colorbar(points,ax=ax[int(np.floor(i/6)),int(i%6)])
    ax[int(np.floor(i/6)),int(i%6)].set_title('Cluster weights in column %s of W'%(str(i)))
plt.show()
```

### 4.6 Strong Cluster Membership Scores

Create a histogram which displays the highest cluster membership score for each of the 500 images.

```python
highest_mem_scores=df_cm_scores_W.T.max()
plt.hist(highest_mem_scores)
plt.title('Highest Membership Score Value for an Image')
plt.xlabel('Membership Score Value')
plt.ylabel('Frequency')
plt.show()
```
